Ñ#´  1 Guía de Contraste de Hipótesis: Comparación de Medias (Prueba t)

Esta guía permite determinar si la diferencia entre dos promedios es **significativa** o si se debe al **azar**, usando el modelo de las empresas de mensajería.

---

##  1. La Sospecha (Planteamiento de Hipótesis)
Definimos qué queremos probar antes de ver los números:
*   **$H_0$ (Hipótesis Nula):** "Las dos empresas son iguales ($\mu_A = \mu_B$). La diferencia es solo suerte".
*   **$H_1$ (Hipótesis Alternativa):** "Existe una diferencia real ($\mu_A \neq \mu_B$). Una es más rápida".

---

##  2. La Evidencia (Tus Datos)
Rellena los paréntesis con la información recolectada:
*   **Empresa A:** Promedio $\bar{x}_A$: `( 2.5 )` | Desviación $s_A$: `( 0.8 )` | Datos $n_A$: `( 30 )`
*   **Empresa B:** Promedio $\bar{x}_B$: `( 2.2 )` | Desviación $s_B$: `( 0.6 )` | Datos $n_B$: `( 30 )`
*   **Diferencia Bruta ($\Delta$):** `( 2.5 )` - `( 2.2 )` = **`( 0.3 )`**

---

##  3. El "Ruido" (Error Estándar)
Calculamos la incertidumbre del experimento usando la fórmula:
$$SE = \sqrt{\frac{s_A^2}{n_A} + \frac{s_B^2}{n_B}}$$

>Recueerda que la Varianza es la desviación estandar s elevada al cuadrado 

*   **Paso A (Var A / n):** `( 0.8 )`² / `( 30 )` = `( 0.0213 )`
*   **Paso B (Var B / n):** `( 0.6 )`² / `( 30 )` = `( 0.0120 )`
*   **Paso C (Suma):** `( 0.0213 )` + `( 0.0120 )` = `( 0.0333 )`
*   **Paso D (Raíz):** $\sqrt{0.0333}$ = **`( 0.1825 )` (ESTE ES TU RUIDO)**

---

##  4. El Duelo Final (Estadístico $t$)
Comparamos la **Diferencia** contra el **Ruido**:
$$t_{calc} = \frac{\bar{x}_A - \bar{x}_B}{SE}$$

*   **Cálculo:** `( 0.3 )` / `( 0.1825 )` = **`( 1.644 )` (TU NÚMERO DE DUELO)**

---

##  5. El Juez (Valor Crítico de Tabla)
Buscamos el límite para un **95% de confianza** ($\alpha = 0.05$):
*   **Grados de Libertad ($df$):** `( 30 )` + `( 30 )` - 2 = **`( 58 )`**
*   **Número Mágico (Tabla $t$):** **`( 2.000 )`**

---

##  6. El Veredicto (Conclusión)
*   **Comparación:** Mi número `( 1.644 )` es **menor** que el del juez `( 2.000 )`.
*   **Resultado:** **No se rechaza $H_0$**.
*   **Conclusión Formal:** "Con un nivel de confianza del 95%, no existe evidencia estadística suficiente para afirmar que hay una diferencia significativa en los tiempos de entrega promedio entre la empresa A y la empresa B".


# 2 Guía de Contraste de Hipótesis: Rendimiento de Software (Muestras Grandes $n > 30$)

Esta guía permite determinar si la mejora en el tiempo de procesamiento entre dos versiones de software es **significativa** o si se debe al **azar**.

---

## 1. La Sospecha (Planteamiento de Hipótesis)
Definimos qué queremos probar antes de ver los números:
*   **$H_0$ (Hipótesis Nula):** "Las dos versiones son iguales ($\mu_1 = \mu_2$). La mejora es solo suerte".
*   **$H_1$ (Hipótesis Alternativa):** "Existe una diferencia real ($\mu_1 \neq \mu_2$). La Versión 2 es más rápida".

---

## 2. La Evidencia (Tus Datos)
Rellena los paréntesis con la información recolectada de las 100 operaciones totales:
*   **Versión 1:** Promedio $\bar{x}_1$: `( 10 )` | Desviación $s_1$: `( 2 )` | Datos $n_1$: `( 50 )`
*   **Versión 2:** Promedio $\bar{x}_2$: `( 9 )` | Desviación $s_2$: `( 1.5 )` | Datos $n_2$: `( 50 )`
*   **Diferencia Bruta ($\Delta$):** `( 10 )` - `( 9 )` = **`( 1.0 )` segundo**

---

## 3. El "Ruido" (Error Estándar)
Calculamos la incertidumbre del experimento usando la fórmula:
$$SE = \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}$$

> Recuerda que la Varianza es la desviación estándar $s$ elevada al cuadrado.

*   **Paso A (Var 1 / n):** `( 2 )`² / `( 50 )` = `( 0.08 )`
*   **Paso B (Var 2 / n):** `( 1.5 )`² / `( 50 )` = `( 0.045 )`
*   **Paso C (Suma):** `( 0.08 )` + `( 0.045 )` = `( 0.125 )`
*   **Paso D (Raíz):** $\sqrt{0.125}$ = **`( 0.3535 )` (ESTE ES TU RUIDO)**

---

## 4. El Duelo Final (Estadístico $Z$)
Comparamos la **Diferencia** contra el **Ruido**:
$$Z_{calc} = \frac{\bar{x}_1 - \bar{x}_2}{SE}$$

*   **Cálculo:** `( 1.0 )` / `( 0.3535 )` = **`( 2.828 )` (TU VALOR Z)**

---

## 5. El Juez (Valor Crítico)
Para un **95% de confianza** ($\alpha = 0.05$) con muestras grandes ($n > 30$):
*   **Número Mágico (Z crítico):** **`( 1.96 )`**

---

## 6. El Veredicto (Conclusión)
*   **Comparación:** Mi número `( 2.828 )` es **mayor** que el del juez `( 1.96 )`.
*   **Resultado:** **SE RECHAZA $H_0$**.
*   **Conclusión Formal:** "Con un nivel de confianza del 95%, existe evidencia estadística suficiente para afirmar que la Versión 2 ofrece un tiempo de procesamiento promedio significativamente mejor que la Versión 1".



# 3 Guía de Contraste de Hipótesis: Comparación de Proporciones (Tratamientos Médicos)

Esta guía permite determinar si la diferencia en las **tasas de recuperación** es significativa o si esa mejora del 5% es solo azar.

---

## 1. La Sospecha (Planteamiento de Hipótesis)
*   **$H_0$ (Hipótesis Nula):** "Ambos tratamientos son igual de efectivos ($p_A = p_B$)."
*   **$H_1$ (Hipótesis Alternativa):** "Existe una diferencia real en la efectividad ($p_A \neq p_B$)."

---

## 2. La Evidencia (Tus Datos)
Convertimos los porcentajes a decimales (75% = 0.75):
*   **Tratamiento A:** Tasa $p_A$: `( 0.75 )` | Fracaso $q_A$: `( 0.25 )` | Datos $n_A$: `( 50 )`
*   **Tratamiento B:** Tasa $p_B$: `( 0.80 )` | Fracaso $q_B$: `( 0.20 )` | Datos $n_B$: `( 50 )`
*   **Diferencia Bruta ($\Delta$):** `( 0.80 )` - `( 0.75 )` = **`( 0.05 )` (5%)**

---

## 3. El "Ruido" (Error Estándar de Proporciones)
Calculamos la incertidumbre. En porcentajes, la varianza es $(p \times q) / n$:
$$SE = \sqrt{\frac{p_A \times q_A}{n_A} + \frac{p_B \times q_B}{n_B}}$$

*   **Paso A (Var A):** `( 0.75 * 0.25 )` / `( 50 )` = `( 0.00375 )`
*   **Paso B (Var B):** `( 0.80 * 0.20 )` / `( 50 )` = `( 0.00320 )`
*   **Paso C (Suma):** `( 0.00375 )` + `( 0.00320 )` = `( 0.00695 )`
*   **Paso D (Raíz):** $\sqrt{0.00695}$ = **`( 0.0833 )` (ESTE ES TU RUIDO)**

---

## 4. El Duelo Final (Estadístico $Z$)
Comparamos la diferencia del 5% contra el ruido:
$$Z_{calc} = \frac{p_B - p_A}{SE}$$

*   **Cálculo:** `( 0.05 )` / `( 0.0833 )` = **`( 0.6002 )` (TU VALOR Z)**

---

## 5. El Juez (Valor Crítico)
Como son 100 pacientes en total (muestra grande), usamos el estándar del 95%:
*   **Número Mágico (Z crítico):** **`( 1.96 )`**

---

## 6. El Veredicto (Conclusión)
*   **Comparación:** Mi número `( 0.6002 )` es **MUCHO MENOR** que el del juez `( 1.96 )`.
*   **Resultado:** **No se rechaza $H_0$**.
*   **Conclusión Formal:** "Con un nivel de confianza del 95%, no existe evidencia suficiente para afirmar que el Tratamiento B es mejor que el A. La diferencia del 5% es pequeña frente al tamaño de la muestra, por lo que podría ser puro azar".


# 3B Guía de Contraste de Hipótesis: Proporciones con Muestra Masiva ($n = 5000$)

Evaluamos si con 10,000 pacientes en total (5000 por grupo), la diferencia del 5% entre el **Tratamiento A (75%)** y el **Tratamiento B (80%)** es estadísticamente significativa.

---

## 1. La Sospecha (Planteamiento de Hipótesis)
*   **$H_0$ (Iguales):** "No hay diferencia real en la efectividad ($\mu_1 = \mu_2$)."
*   **$H_1$ (Diferentes):** "El Tratamiento B es significativamente más efectivo ($\mu_1 \neq \mu_2$)."

---

## 2. La Evidencia (Tus Datos)
*   **Tratamiento A:** Tasa $p_A$: `( 0.75 )` | Fracaso $q_A$: `( 0.25 )` | Datos $n_A$: `( 5000 )`
*   **Tratamiento B:** Tasa $p_B$: `( 0.80 )` | Fracaso $q_B$: `( 0.20 )` | Datos $n_B$: `( 5000 )`
*   **Diferencia Bruta ($\Delta$):** `( 0.80 )` - `( 0.75 )` = **`( 0.05 )` (5%)**

---

## 3. El "Ruido" (Error Estándar de Proporciones)
Con 5000 pacientes, la incertidumbre casi desaparece:
$$SE = \sqrt{\frac{p_A \times q_A}{n_A} + \frac{p_B \times q_B}{n_B}}$$

*   **Paso A (Var A / n):** `( 0.75 * 0.25 )` / `( 5000 )` = `( 0.0000375 )`
*   **Paso B (Var B / n):** `( 0.80 * 0.20 )` / `( 5000 )` = `( 0.0000320 )`
*   **Paso C (Suma):** `( 0.0000375 )` + `( 0.0000320 )` = `( 0.0000695 )`
*   **Paso D (Raíz):** $\sqrt{0.0000695}$ = **`( 0.00833 )` (ESTE ES TU RUIDO MINIATURA)**

---

## 4. El Duelo Final (Estadístico $Z$)
Comparamos la diferencia del 5% contra este ruido tan pequeñito:
$$Z_{calc} = \frac{0.05}{0.00833}$$

*   **Cálculo:** `( 0.05 )` / `( 0.00833 )` = **`( 6.002 )` (TU VALOR Z GIGANTE)**

---

## 5. El Juez (Valor Crítico)
*   **Confianza:** 95%
*   **Número Mágico (Z crítico):** **`( 1.96 )`**

---

## 6. El Veredicto (Conclusión)
*   **Comparación:** Mi número `( 6.002 )` es **MUCHO MAYOR** que el del juez `( 1.96 )`.
*   **Resultado:** **SE RECHAZA $H_0$ TOTALMENTE**.
*   **Conclusión Formal:** "Con un 95% de confianza, se afirma que el Tratamiento B es significativamente más efectivo que el Tratamiento A. La probabilidad de que este resultado sea por azar es casi nula (p < 0.001)".

###  ¿Cómo elegir el número correcto según la colita y la cantidad de valores ( grados de libertad) ?

1.  **Mira tu $H_1$:** 
    *   Si usaste el signo **$\neq$**, busca en la fila de **2 Colas**.
    *   Si usaste los signos **$>$** o **$<$**, busca en la fila de **1 Cola**.
2.  **Elige tu seguridad:** Lo estándar en el mundo científico y financiero es el **95% de confianza**.
3.  **El Veredicto:** 
    *   Si tu resultado ($Z$ calculado) es **MAYOR** que el número de la tabla $\rightarrow$ **¡Victoria!** Hay diferencia significativa.
    *   Si es **MENOR** $\rightarrow$ **Empate técnico.** La diferencia se debe al azar.

> **Nota para T-Student ($n < 30$):** Si tienes pocos datos, estos números "Z" no te sirven. Debes buscar en la **Tabla T** usando tus Grados de Libertad ($df$).
