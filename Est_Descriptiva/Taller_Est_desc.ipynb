#  Ejemplo: Control de Calidad (Cajas de Helado 500g)

En este ejemplo, transformaremos 30 mediciones desordenadas en información valiosa para la toma de decisiones.

---

###  PASO 0: Los Datos Crudos (n = 30)
Primero, organizamos los datos de menor a mayor para facilitar el conteo:
**494, 495, 496, 497, 497, 497, 498, 498, 498, 499, 499, 499, 499, 500, 500, 500, 500, 501, 501, 501, 502, 502, 502, 503, 503, 503, 504, 504, 505, 506**

---

###  PASO 1: Organización (Tabla de Frecuencias)

Para construir la tabla, definimos:
1. **Rango ($R$):** $Dato_{max} - Dato_{min} = 506 - 494 = 12$.

###  Paso 1.1: Cálculo del Número de Intervalos ($k$)

Para no elegir el número de filas al azar, utilizamos la **Regla de Sturges**, que es la fórmula estándar en estadística para organizar datos:

$$k = 1 + 3.322 \cdot \log_{10}(n)$$

**Donde:**
*   **$n$**: Es el número total de datos ($n = 30$).
*   **$\log_{10}$**: Logaritmo en base 10 (puedes usar la calculadora o Excel con `=LOG10(30)`).

**Operación para nuestro ejemplo:**
1.  $\log_{10}(30) \approx 1.477$
2.  $3.322 \cdot 1.477 \approx 4.90$
3.  $1 + 4.90 = 5.90$
4.  **Redondeo:** Como el resultado es $5.90$, lo aproximamos al entero más cercano: **$k = 6$**.

---

###  Paso 1.2: Cálculo de la Amplitud ($A$)

La amplitud nos dice qué tan "ancho" es cada intervalo (cuántos gramos abarca cada fila de la tabla). Se calcula dividiendo el **Rango ($R$)** entre el **Número de intervalos ($k$)**:

$$A = \frac{R}{k}$$

**Operaciones:**
1.  **Rango ($R$):** $Dato_{max} - Dato_{min} = 506 - 494 = 12$.
2.  **Intervalos ($k$):** Usaremos **6** intervalos para cubrir el rango.
3.  **Amplitud ($A$):** $12 / 6 = 2$.

> **Resultado:** Nuestra tabla tendrá **6 filas** y cada una avanzará de **2 en 2** gramos (por ejemplo: 494 a 496, 496 a 498...).



2. **Intervalos ($k$):** Usaremos **6** intervalos para cubrir el rango.
3. **Amplitud ($A$):** $R / k = 12 / 6 = 2$. Cada intervalo saltará de 2 en 2.

###  PASO 2: Medidas de Tendencia Central (¿Dónde está el centro?)

1. **Media ($\bar{x}$):** Es el promedio. Sumamos todos los datos y dividimos por 30.
   $$\bar{x} = \frac{\sum x_i}{n} = \frac{15006}{30} = 500.2 \text{ g}$$
   *Interpretación: El peso promedio de las cajas es casi perfecto (500.2g).*

2. **Mediana ($\tilde{x}$):** Es el valor central. En 30 datos, es el promedio de las posiciones 15 y 16.
   $$\tilde{x} = \frac{500 + 500}{2} = 500 \text{ g}$$

3. **Moda ($Mo$):** Los valores que más se repiten.
   *   **499** y **500** aparecen 4 veces cada uno. Es una distribución **Bimodal**.

###  PASO 3: Medidas de Dispersión (¿Qué tan variable es?)

### 📉 Cálculo de la Varianza desde la Tabla de Frecuencias

Cuando los datos están agrupados, no necesitamos restar cada dato individual. Usamos la **Marca de Clase ($x_i$)** y la multiplicamos por su **Frecuencia ($f_i$)**.

#### Paso 1: Tabla Auxiliar de Dispersión

---

###  Aplicando la Fórmula Grupal

La fórmula de la varianza para datos agrupados es:

$$s^2 = \frac{\sum f_i \cdot (x_i - \bar{x})^2}{n - 1}$$

**Sustituyendo nuestros valores:**

1. **Suma de desviaciones pesadas:** $248.8$
2. **Dividir por $n-1$:**
   $$s^2 = \frac{248.8}{29} \approx 8.58$$
3. **Desviación Estándar ($s$):**
   $$s = \sqrt{8.58} \approx 2.93 \text{ g}$$

---

###  Aplicando la Fórmula con Cálculo Individual

   $$s^2 = \frac{\sum (x_i - \bar{x})^2}{n - 1} = \frac{280.8}{29} \approx 9.68$$

2. **Desviación Estándar ($s$):** Es la raíz cuadrada de la varianza. Nos da la desviación en las mismas unidades que los datos (gramos).
   $$s = \sqrt{9.68} \approx 3.11 \text{ g}$$
   *Interpretación: En promedio, las cajas se desvían $\pm 3.11$ g del peso central.*

###  PASO 4: Forma y Posición

### 1. Análisis de la Curtosis (Forma de la Curva)

La **Curtosis** nos indica qué tan concentrados están los datos alrededor de la media. Nos dice si la "campana" es alta y delgada o bajita y ancha.

---

#### 1. Tipos de Curtosis (¿Cómo se ve mi gráfica?)

Dependiendo del resultado del coeficiente ($K$), clasificamos la curva en tres tipos:

1.  **Leptocúrtica ($K > 0$):** Curva muy puntiaguda. Los datos están muy amontonados en el centro.
2.  **Mesocúrtica ($K = 0$):** Es la "Distribución Normal" perfecta. Ni muy alta, ni muy plana.
3.  **Platicúrtica ($K < 0$):** Curva achatada o plana. Los datos están más dispersos hacia los extremos (colas).

---

### 2: Cálculo de la Curtosis (Datos Agrupados)

Para calcular la forma de la campana sin procesar los 30 datos uno por uno, extendemos nuestra **Tabla de Frecuencias** con una columna de "Potencia Cuarta".

#### 1. Tabla de Frecuencias Extendida
Recordamos nuestra media ($\bar{x} = 500.2$) y nuestra desviación estándar ($s \approx 2.93$).

| :--- | :---: | :---: | :---: | :---: | :---: |
| [494 – 496) | 495 | 2 | $27.04$ | $54.08$ | $1,462.3$ |
| [496 – 498) | 497 | 4 | $10.24$ | $40.96$ | $419.4$ |

| [502 – 504) | 503 | 6 | $7.84$ | $47.04$ | $368.8$ |
| [504 – 506] | 505 | 4 | $23.04$ | $92.16$ | $2,123.3$ |
| **TOTAL** | | **30** | | **$\sum = 248.8$** | **$\sum = 4,391.1$** |

---

#### 4. La Fórmula de Curtosis para Datos Agrupados

Utilizamos el **Momento de Orden 4** ($g_2$). La fórmula para datos agrupados es:

$$K = \left[ \frac{\sum f_i \cdot (x_i - \bar{x})^4}{n \cdot s^4} \right] - 3$$

**Sustituyendo los valores de nuestra tabla:**

1.  **Suma de potencias cuartas ($\sum$):** $4,391.1$
2.  **Denominador ($n \cdot s^4$):** $30 \cdot (2.93)^4 \approx 30 \cdot 73.7 \approx 2,211$
3.  **Cálculo Final:**
    $$K = \frac{4,391.1}{2,211} - 3 \approx 1.98 - 3 = -1.02$$

---

#### 5. Interpretación de los Resultados

*   **Resultado ($K \approx -1.02$):** Como el valor es **negativo**, confirmamos que la distribución es **Platicúrtica**.
*   **Significado:** La máquina de helados tiene una producción "achatada". No hay un pico exagerado en los 500g, sino que los pesos están distribuidos de forma más uniforme y plana a lo largo de los intervalos. 

> **Tip para Excel:** Aunque lo hagas a mano, siempre puedes verificar seleccionando tus datos y usando `=CURTOSIS(rango)`.
