##  PARTE 1: Análisis del Departamento A (30 empleados)

Primero, construimos nuestra tabla de frecuencias. Agruparemos las notas en intervalos de 3 en 3.
Añadimos una columna "mágica" llamada **$x_i \cdot f_i$**. Esta es simplemente la multiplicación del centro del grupo (Marca de clase) por la cantidad de personas en ese grupo. Nos servirá para calcular el promedio.

| Intervalo (Notas) | Marca de Clase ($x_i$) | Frecuencia ($f_i$) | Frecuencia Acumulada ($F_i$) | Multiplicación ($x_i \cdot f_i$) |
| :--- | :--- | :--- | :--- | :--- |
| **85 – 87** | 86 | **5** | **5** | 86 × 5 = **430** |
| **88 – 90** | 89 | **13** | 5 + 13 = **18** | 89 × 13 = **1157** |
| **91 – 93** | 92 | **9** | 18 + 9 = **27** | 92 × 9 = **828** |
| **94 – 96** | 95 | **3** | 27 + 3 = **30** | 95 × 3 = **285** |
| **TOTAL** | | **$n = 30$** | | **Suma = 2700** |

### 1. La Media (El Promedio)
Cuando los datos están agrupados, asumimos que todos los empleados de un intervalo sacaron la nota del medio ($x_i$). 
*   **Fórmula:** Dividimos la suma total de la última columna entre el número de empleados.
*   **Cálculo:** Media = $2700 \div 30$ = **90**.
*   *Interpretación:* El rendimiento promedio del Departamento A es 90.

### 2. La Mediana (La nota del medio)
Tenemos 30 empleados. La mitad es 15 ($n/2 = 15$). ¿En qué grupo está el empleado número 15? 
Miramos la Frecuencia Acumulada ($F_i$). El primer grupo llega hasta el empleado 5. El segundo grupo llega hasta el 18. ¡Ajá! El empleado 15 está en el intervalo **88 - 90**. Este es nuestro "Intervalo Mediano".

Para ser exactos, usamos una fórmula de detectives.
*   **Límite inferior exacto ($L_i$):** El intervalo empieza en 88, pero en matemáticas exactas, su frontera con el grupo anterior es **87.5**.
*   **Amplitud ($A$):** El tamaño del salto de un grupo a otro es **3**.
*   **Fórmula:** $Mediana = L_i + \left[ \frac{15 - F_{Anterior}}{f_{Mediana}} \right] \cdot A$
*   **Cálculo:** $Mediana = 87.5 + \left[ \frac{15 - 5}{13} \right] \cdot 3$
*   $Mediana = 87.5 + \left[ \frac{10}{13} \right] \cdot 3$ = $87.5 + (0.769 \cdot 3)$ = $87.5 + 2.31$ = **89.81**.

### 3. La Moda (La nota más popular)
Buscamos el grupo con más personas (mayor $f_i$). Es el grupo **88 - 90**, que tiene 13 personas.
Para estimar el punto exacto más de moda dentro de ese grupo, comparamos su tamaño con sus vecinos usando la fórmula de la moda para datos agrupados:
*   **Diferencia 1 ($d_1$):** Frecuencia de la moda menos el grupo de arriba (13 - 5 = **8**)
*   **Diferencia 2 ($d_2$):** Frecuencia de la moda menos el grupo de abajo (13 - 9 = **4**)
*   **Fórmula:** $Moda = L_i + \left[ \frac{d_1}{d_1 + d_2} \right] \cdot A$
*   **Cálculo:** $Moda = 87.5 + \left[ \frac{8}{8 + 4} \right] \cdot 3$ = $87.5 + \left[ \frac{8}{12} \right] \cdot 3$ = $87.5 + 2$ = **89.5**.

##  PARTE 3: Medidas de Dispersión en Datos Agrupados

Ya sabemos que el Promedio del Departamento A es 90 y el del B es 73.43. Pero, ¿qué tan parecidas son las notas entre los empleados de un mismo departamento? ¿Están todos cerca del promedio o hay notas muy altas y muy bajas mezcladas? 

Para averiguarlo, calcularemos la **Varianza** y la **Desviación Estándar** usando nuestras tablas. 

**El Truco del Detective:** 
Vamos a calcular la "distancia" de cada grupo hasta el promedio, la elevaremos al cuadrado (para quitar signos negativos) y la multiplicaremos por la cantidad de personas en ese grupo. 
La fórmula es: **$f_i \cdot (x_i - Promedio)^2$**

---

### Dispersión del Departamento A
Recordemos: El promedio de este grupo nos dio **90**. El total de empleados es $n = 30$.

Vamos a ver qué tan lejos está el centro de cada grupo ($x_i$) de ese hermoso 90:

| Marca de Clase ($x_i$) | Frecuencia ($f_i$) | 1. Restamos el Promedio | 2. Al cuadrado | 3. Multiplicamos por $f_i$ |
| :--- | :--- | :--- | :--- | :--- |
| **86** | **5** | $86 - 90 = -4$ | $(-4)^2 = 16$ | $16 \times 5 = \mathbf{80}$ |
| **89** | **13** | $89 - 90 = -1$ | $(-1)^2 = 1$ | $1 \times 13 = \mathbf{13}$ |
| **92** | **9** | $92 - 90 = 2$ | $(2)^2 = 4$ | $4 \times 9 = \mathbf{36}$ |
| **95** | **3** | $95 - 90 = 5$ | $(5)^2 = 25$ | $25 \times 3 = \mathbf{75}$ |
| | | | | **Suma Total = 204** |

1. **Varianza ($S^2$):** Tomamos esa "Suma Total" y la dividimos entre el total de empleados menos uno ($n - 1$, es decir, 29).
   * **Cálculo:** $204 \div 29$ = **7.03**

2. **Desviación Estándar ($S$):** Es simplemente la raíz cuadrada de la varianza. Nos devuelve el valor a los "puntos" originales.
   * **Cálculo:** $\sqrt{7.03}$ = **2.65 puntos**.
   * *¿Qué significa?* En promedio, las notas del Departamento A se alejan unos 2.65 puntos hacia arriba o hacia abajo del gran promedio de 90.

---

### Dispersión del Departamento B
Recordemos: El promedio exacto de este grupo nos dio **73.43**. El total de empleados es $n = 30$.

Hagamos la misma magia de las distancias:

| Marca de Clase ($x_i$) | Frecuencia ($f_i$) | 1. Restamos el Promedio | 2. Al cuadrado | 3. Multiplicamos por $f_i$ |
| :--- | :--- | :--- | :--- | :--- |
| **70.5** | **5** | $70.5 - 73.43 = -2.93$ | $(-2.93)^2 = 8.58$ | $8.58 \times 5 = \mathbf{42.92}$ |
| **72.5** | **10** | $72.5 - 73.43 = -0.93$ | $(-0.93)^2 = 0.86$ | $0.86 \times 10 = \mathbf{8.65}$ |
| **74.5** | **11** | $74.5 - 73.43 = 1.07$ | $(1.07)^2 = 1.14$ | $1.14 \times 11 = \mathbf{12.59}$ |
| **76.5** | **4** | $76.5 - 73.43 = 3.07$ | $(3.07)^2 = 9.42$ | $9.42 \times 4 = \mathbf{37.70}$ |
| | | | | **Suma Total = 101.86** |

1. **Varianza ($S^2$):** Dividimos la suma total entre 29 ($n - 1$).
   * **Cálculo:** $101.86 \div 29$ = **3.51**

2. **Desviación Estándar ($S$):** Sacamos la raíz cuadrada de la varianza.
   * **Cálculo:** $\sqrt{3.51}$ = **1.87 puntos**.
   * *¿Qué significa?* Las notas del Departamento B se alejan en promedio solo 1.87 puntos de su nota central (73.43).

---
**🏆 El Veredicto Final de la Dispersión:**
¡El Departamento B tiene una Desviación Estándar menor (1.87 vs 2.65)! Esto significa que los empleados del Departamento B son más **consistentes** o "parejos" entre ellos. Aunque sus notas son más bajas en general, casi todos sacan notas muy similares. 

Por otro lado, en el Departamento A hay mayor desigualdad: las notas están más separadas de su propio promedio. 

*(Dato curioso para mentes brillantes: Si comparas estos resultados con los datos originales sin agrupar, verás que varían un poquitito. ¡A eso se le llama "error de agrupamiento", y ocurre porque al agrupar datos sacrificamos un poquito de precisión para ganar mucha velocidad!)*


##  PARTE 4: La Prueba t de Student (¿Es la diferencia REAL? con dos colitas) 

Hemos visto que el Departamento A tiene un promedio más alto y el B uno más bajo. Pero, ¿es esta diferencia lo suficientemente grande como para decir que el Departamento A es *realmente* mejor, o podría ser que, si volviéramos a medir, los resultados fueran diferentes por pura suerte?

Para responder a esta pregunta, usamos una herramienta estadística muy poderosa: la **Prueba t de Student**.

### ¿Qué necesitamos para nuestra "receta" de la Prueba t?

Vamos a usar los números que ya calculamos de nuestras tablas de frecuencias:

*   **Departamento A:**
    *   Promedio ($\bar{x}_A$) = **90**
    *   Varianza ($S^2_A$) = **7.03**
    *   Número de empleados ($n_A$) = **30**

*   **Departamento B:**
    *   Promedio ($\bar{x}_B$) = **73.43**
    *   Varianza ($S^2_B$) = **3.51**
    *   Número de empleados ($n_B$) = **30**

### La Gran Pregunta (Hipótesis):

En estadística, siempre empezamos con dos ideas opuestas:

*   **Hipótesis Nula (H₀):** "No hay diferencia". Esto significa que, en realidad, los promedios de los dos departamentos son iguales. Cualquier diferencia que vemos es solo por casualidad.
*   **Hipótesis Alternativa (H₁):** "Sí hay diferencia". Esto significa que el promedio de un departamento es realmente diferente al del otro.

### La Fórmula de la Prueba t (¡No te asustes, la desglosaremos!)

La fórmula que usaremos para calcular nuestro valor "t" es:

$t = \frac{\text{Diferencia entre los promedios}}{\text{Error estándar de la diferencia}}$

O, con símbolos:

$t = \frac{\bar{x}_A - \bar{x}_B}{\sqrt{\frac{S^2_A}{n_A} + \frac{S^2_B}{n_B}}}$

### ¡Manos a la Obra! Calculando nuestro valor "t" paso a paso:

1.  **Paso 1: La diferencia entre los promedios (la parte de arriba de la fórmula)**
    *   Restamos el promedio del Departamento B al del Departamento A:
    *   $90 - 73.43 = \mathbf{16.57}$

2.  **Paso 2: Calculando el "Error Estándar de la Diferencia" (la parte de abajo de la fórmula)**
    *   Esta parte nos dice qué tan "ruidosa" o variable es la diferencia entre los promedios.
    *   Primero, dividimos la varianza de cada departamento entre su número de empleados:
        *   Para el Departamento A: $7.03 \div 30 = \mathbf{0.2343}$
        *   Para el Departamento B: $3.51 \div 30 = \mathbf{0.117}$
    *   Luego, sumamos esos dos resultados:
        *   $0.2343 + 0.117 = \mathbf{0.3513}$
    *   Finalmente, sacamos la raíz cuadrada de esa suma:
        *   $\sqrt{0.3513} \approx \mathbf{0.5927}$

3.  **Paso 3: Calculando el valor "t" final**
    *   Ahora dividimos el resultado del Paso 1 entre el resultado del Paso 2:
    *   $t = 16.57 \div 0.5927 \approx \mathbf{27.96}$

### El Veredicto Final: ¿Qué significa nuestro valor "t"?

Nuestro valor "t" es **27.96**. ¿Es esto un número grande o pequeño?

En estadística, para saber si una diferencia es "real" (estadísticamente significativa) con un nivel de confianza común, usamos una regla general para estudiantes de noveno grado:

*   **Si el valor absoluto de "t" es mayor que 2**, podemos decir que la diferencia es **estadísticamente significativa**. Esto significa que es muy poco probable que la diferencia que observamos sea solo por casualidad.

**Nuestra Conclusión:**

Dado que nuestro valor "t" (27.96) es **MUCHO mayor que 1.96**, podemos **rechazar la Hipótesis Nula** en dos colitas.

Esto significa que tenemos una evidencia muy fuerte para decir que **existe una diferencia estadísticamente significativa** entre el rendimiento promedio del Departamento A y el Departamento B. ¡El Departamento A realmente tiene un rendimiento superior!

1.  **Tu valor "t" calculado:** En nuestro caso, fue **27.96**.
2.  **Los "Grados de Libertad" (gl):** Es una medida que se relaciona con cuántos datos tienes. Para una prueba t de dos muestras como la nuestra, una forma sencilla de aproximarlo es:
    *   $gl = (n_A - 1) + (n_B - 1) = (30 - 1) + (30 - 1) = 29 + 29 = \mathbf{58}$

```python
import scipy.stats as stats

# Nuestros datos originales
data_a =
data_b =

# Realizamos la prueba t para muestras independientes.
# equal_var=False es importante porque ya vimos que las varianzas no eran iguales.
t_statistic, p_value_two_tails = stats.ttest_ind(data_a, data_b, equal_var=False)

# Para una prueba de una cola (si A > B)
p_value_one_tail = p_value_two_tails / 2

print(f"El valor 't' calculado es: {t_statistic:.2f}")
print(f"P-valor para una prueba de dos colas: {p_value_two_tails}")
print(f"P-valor para una prueba de una cola (A > B): {p_value_one_tail}")


##  PARTE 5: La Prueba t de Student para "Mayor Que" (Prueba de Una Cola)

En la Parte 4, hicimos una prueba de "dos colas" porque solo queríamos saber si había *alguna* diferencia entre los departamentos (sin importar cuál era mejor). Pero, ¿qué pasa si nuestra pregunta inicial era más específica? Por ejemplo:

**"¿El rendimiento promedio del Departamento A es significativamente MAYOR que el del Departamento B?"**

Cuando tenemos una dirección específica ("mayor que" o "menor que"), realizamos una **Prueba de una cola**.

### 1. Las Nuevas Hipótesis (¡Ahora con dirección!)

Nuestras ideas opuestas cambian un poco para reflejar esta dirección:

*   **Hipótesis Nula (H₀):** El promedio del Departamento A **NO es mayor** que el del Departamento B. (Es decir, $\bar{x}_A \le \bar{x}_B$).
    *   *En palabras sencillas:* El Departamento A no rinde mejor que el B, o rinde igual, o incluso peor.
*   **Hipótesis Alternativa (H₁):** El promedio del Departamento A **ES mayor** que el del Departamento B. (Es decir, $\bar{x}_A > \bar{x}_B$).
    *   *En palabras sencillas:* El Departamento A rinde significativamente mejor que el B.

### 2. El Cálculo del Valor "t" (¡Es el mismo!)

La buena noticia es que el cálculo del valor "t" **no cambia**. La fórmula y los números son los mismos que usamos en la prueba de dos colas:

$t = \frac{\bar{x}_A - \bar{x}_B}{\sqrt{\frac{S^2_A}{n_A} + \frac{S^2_B}{n_B}}}$

Como ya calculamos:

*   Nuestro valor "t" es aproximadamente **27.96**.

### 3. El Veredicto Final para "Mayor Que" (¡Aquí está la diferencia clave!)

La principal diferencia con la prueba de dos colas está en cómo interpretamos este valor "t" y, más importante, el **p-valor**.

*   **P-valor:** Es la probabilidad de obtener un resultado tan extremo (o más extremo) como el que observamos, *asumiendo que la Hipótesis Nula es verdadera*.
    *   Para una prueba de una cola, si la dirección de nuestra diferencia observada (A es mayor que B) coincide con nuestra hipótesis alternativa (A es mayor que B), el p-valor de una cola es aproximadamente **la mitad** del p-valor de una prueba de dos colas.
    *   En nuestro caso anterior (prueba de dos colas), el p-valor era extremadamente pequeño (casi cero). Si fuera, por ejemplo, 0.000000000000000000000000000133, para una prueba de una cola en la dirección correcta, sería la mitad de eso.

*   **Regla de Decisión:**
    *   Si el p-valor de una cola es **menor** que nuestro nivel de significancia (generalmente 0.05), entonces **rechazamos la Hipótesis Nula**.
    *   Si el p-valor es **mayor** que 0.05, **no podemos rechazar la Hipótesis Nula**.

**Nuestra Conclusión para "Mayor Que":**

Dado que nuestro valor "t" (27.96) es muy grande y positivo, y la diferencia observada (A es mucho mayor que B) va en la dirección de nuestra hipótesis alternativa (A > B), el p-valor de una cola será **extremadamente pequeño** (mucho menor que 0.05).

Por lo tanto, **rechazamos la Hipótesis Nula**.

Esto significa que tenemos una evidencia abrumadora para concluir que **el rendimiento promedio del Departamento A es significativamente MAYOR que el del Departamento B**.

---
**¡Un paso más allá!** Entender la diferencia entre pruebas de una y dos colas te permite hacer preguntas más precisas a tus datos y obtener respuestas más directas.
