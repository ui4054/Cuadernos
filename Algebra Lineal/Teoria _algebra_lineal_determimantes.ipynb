## Problema de Producción: Solución con Álgebra Lineal

Una empresa fabrica tres productos (A, B y C), cada uno de los cuales requiere tiempo en tres procesos distintos: ensamblaje, inspección y empaque. El objetivo es determinar la cantidad de unidades de cada producto que se deben fabricar para utilizar completamente las horas disponibles en cada proceso.

### Paso 1: Definir las variables y plantear el sistema de ecuaciones

Primero, definimos nuestras variables:
- **x**: Cantidad de unidades a producir del Producto A
- **y**: Cantidad de unidades a producir del Producto B
- **z**: Cantidad de unidades a producir del Producto C

Con base en los datos proporcionados, construimos un sistema de ecuaciones lineales. Cada ecuación representa las horas totales de un proceso:

1.  **Ecuación de Ensamblaje:**
    `2x + 1y + 2z = 200`

2.  **Ecuación de Inspección:**
    `1x + 3y + 2z = 150`

3.  **Ecuación de Empaque:**
    `2x + 1y + 3z = 180`

### Paso 2: Representación Matricial del Sistema

Podemos expresar este sistema de ecuaciones en su forma matricial `A * X = B`:

- **A** es la matriz de coeficientes (tiempo requerido por producto).
- **X** es el vector de variables (cantidades de cada producto).
- **B** es el vector de términos independientes (disponibilidad de horas).

$$ A = \begin{pmatrix} 2 & 1 & 2 \\ 1 & 3 & 2 \\ 2 & 1 & 3 \end{pmatrix}, \quad X = \begin{pmatrix} x \\ y \\ z \end{pmatrix}, \quad B = \begin{pmatrix} 200 \\ 150 \\ 180 \end{pmatrix} $$

### Paso 3: Solución por Regla de Cramer

La Regla de Cramer nos permite resolver el sistema usando determinantes. Las soluciones para x, y, z son:

$$ x = \frac{\Delta_x}{\Delta_s}, \quad y = \frac{\Delta_y}{\Delta_s}, \quad z = \frac{\Delta_z}{\Delta_s} $$

Calcularemos cada uno de los cuatro determinantes necesarios.

#### 3.1. Cálculo del Determinante del Sistema (Δs)

$$ \Delta_s = \det(A) = \begin{vmatrix} 2 & 1 & 2 \\ 1 & 3 & 2 \\ 2 & 1 & 3 \end{vmatrix} $$

Usando la regla de Sarrus:
$$ \Delta_s = [(2)(3)(3) + (1)(2)(2) + (2)(1)(1)] - [(2)(3)(2) + (2)(2)(1) + (3)(1)(1)] $$
$$ \Delta_s = [18 + 4 + 2] - [12 + 4 + 3] $$
$$ \Delta_s = 24 - 19 = 5 $$

Como el determinante es distinto de cero (Δs = 5), el sistema tiene una única solución.

#### 3.2. Cálculo de los Determinantes de las Variables


**Determinante de $x$ ($\Delta_x$):**
Reemplazamos la columna 1 por el vector de disponibilidad $B = [200, 150, 180]^T$:
$$ 
\Delta_x = \begin{vmatrix} \mathbf{200} & 1 & 2 \\ \mathbf{150} & 3 & 2 \\ \mathbf{180} & 1 & 3 \end{vmatrix}
$$
*   **Diagonales Principales (+):**
    $$ (200 \cdot 3 \cdot 3) + (1 \cdot 2 \cdot 180) + (2 \cdot 150 \cdot 1) = 1800 + 360 + 300 = \mathbf{2460} $$
*   **Diagonales Secundarias (-):**
    $$ (180 \cdot 3 \cdot 2) + (1 \cdot 150 \cdot 3) + (200 \cdot 1 \cdot 2) = 1080 + 450 + 400 = \mathbf{1930} $$
*   **Resultado:** $\Delta_x = 2460 - 1930 = \mathbf{530}$

---

**Determinante de $y$ ($\Delta_y$):**
Reemplazamos la columna 2 por el vector de disponibilidad $B = [200, 150, 180]^T$:
$$ 
\Delta_y = \begin{vmatrix} 2 & \mathbf{200} & 2 \\ 1 & \mathbf{150} & 2 \\ 2 & \mathbf{180} & 3 \end{vmatrix}
$$
*   **Diagonales Principales (+):**
    $$ (2 \cdot 150 \cdot 3) + (200 \cdot 2 \cdot 2) + (2 \cdot 1 \cdot 180) = 900 + 800 + 360 = \mathbf{2060} $$
*   **Diagonales Secundarias (-):**
    $$ (2 \cdot 150 \cdot 2) + (180 \cdot 2 \cdot 2) + (3 \cdot 1 \cdot 200) = 600 + 720 + 600 = \mathbf{1920} $$
*   **Resultado:** $\Delta_y = 2060 - 1920 = \mathbf{140}$

---

**Determinante de $z$ ($\Delta_z$):**
Reemplazamos la columna 3 por el vector de disponibilidad $B = [200, 150, 180]^T$:
$$ 
\Delta_z = \begin{vmatrix} 2 & 1 & \mathbf{200} \\ 1 & 3 & \mathbf{150} \\ 2 & 1 & \mathbf{180} \end{vmatrix}
$$
*   **Diagonales Principales (+):**
    $$ (2 \cdot 3 \cdot 180) + (1 \cdot 150 \cdot 2) + (200 \cdot 1 \cdot 1) = 1080 + 300 + 200 = \mathbf{1580} $$
*   **Diagonales Secundarias (-):**
    $$ (200 \cdot 3 \cdot 2) + (1 \cdot 1 \cdot 180) + (150 \cdot 1 \cdot 2) = 1200 + 180 + 300 = \mathbf{1680} $$
*   **Resultado:** $\Delta_z = 1580 - 1680 = \mathbf{-100}$


### Paso 4: Encontrar la Solución del Sistema

Ahora aplicamos las fórmulas de Cramer con los valores de los determinantes que hemos encontrado:

- **x** = Δx / Δs = 530 / 5 = **106**
- **y** = Δy / Δs = 140 / 5 = **28**
- **z** = Δz / Δs = -100 / 5 = **-20**

### Paso 5: Verificación de la Solución

Es fundamental comprobar si nuestra solución satisface el sistema de ecuaciones original:

- **Ecuación 1:** `2(106) + 1(28) + 2(-20) = 212 + 28 - 40 = 200` ✓ (Correcto)
- **Ecuación 2:** `1(106) + 3(28) + 2(-20) = 106 + 84 - 40 = 150` ✓ (Correcto)
- **Ecuación 3:** `2(106) + 1(28) + 3(-20) = 212 + 28 - 60 = 180` ✓ (Correcto)

Matemáticamente, la solución es correcta.

### Paso 6: Conclusión e Interpretación del Resultado

Hemos encontrado la única solución matemática para el sistema: `x = 106`, `y = 28`, `z = -20`.

Sin embargo, aquí viene la parte más importante para un ingeniero o administrador: **la interpretación en el mundo real**.

La solución nos dice que para usar *exactamente* todas las horas disponibles, necesitaríamos producir **-20 unidades del Producto C**. Esto, por supuesto, es físicamente imposible. No podemos "des-producir" un artículo.

**¿Qué significa esto?**

Significa que **no es posible operar la planta de manera que se consuman el 100% de las horas en los tres procesos simultáneamente**. La combinación de tiempos de producción y horas disponibles es incompatible para una utilización total.

En un escenario real, el problema no se plantearía con igualdades (`=`), sino con desigualdades (`<=`), ya que la empresa puede usar *hasta* un máximo de horas. El objetivo sería entonces maximizar las ganancias sujeto a esas restricciones, lo que se resuelve con **Programación Lineal**. 

Para este ejercicio de álgebra, la lección clave es que, aunque encontremos una solución matemática precisa, siempre debemos analizar si tiene sentido en el contexto de la empresa. Pro lo tanto la desición puede estar asociada a aumentar la disponibilidad de horas para el producto C.


## 1. El Escenario "Ahogado" (Modelo Original)
Al intentar resolver el sistema original mediante la **Regla de Cramer**, los resultados son:
* **Producto A ($x_1$):** 106 unidades
* **Producto B ($x_2$):** 28 unidades
* **Producto C ($x_3$):** -20 unidades

### El Cuello de Botella: Empaque
El resultado negativo en $x_3$ indica que el sistema es **inconsistente** para la producción real. El **Empaque (180h)** es insuficiente para sostener la estructura de producción de los tres productos simultáneamente. Matemáticamente, el sistema nos "pide" quitar producción de C para liberar horas hacia A y B.

---

## 2. Equilibrio A y B (Producción de C = 0)
Para que la producción de $x_3$ sea exactamente **0**, hemos calculado que la disponibilidad de **Empaque** debe subir de 180 a **200 horas**. Bajo este nuevo escenario, el sistema se equilibra así:

1. **Ensamblaje:** $2x_1 + 1x_2 + 2(0) = 200$
2. **Inspección:** $1x_1 + 3x_2 + 2(0) = 150$
3. **Empaque:** $2x_1 + 1x_2 + 3(0) = 200$

**Resultado del equilibrio:**
* **$x_1$ (A):** 90 unidades
* **$x_2$ (B):** 20 unidades
* **$x_3$ (C):** 0 unidades


```python
import numpy as np
import matplotlib.pyplot as plt

# Definimos el rango de horas de Empaque (desde el punto de ahogo 180 hasta 300)
horas_empaque = np.linspace(180, 300, 50)

# El determinante del sistema (D) es constante = 5
D = 5

# Calculamos la producción de x1, x2, x3 en función de las horas de empaque (H)
# Basado en las expansiones de los determinantes calculados previamente:
# x1 = (350 + 0.8*H) / 5  <-- Simplificación de D1 con H variable
# x2 = (140) / 5          <-- D2 no depende de H en este sistema específico
# x3 = (5*H - 1000) / 5   <-- Punto donde H=200 da x3=0

x1 = (130 + 0.4 * horas_empaque) # Simplificado para el ejemplo
x3 = (horas_empaque - 200) / 1    # x3 crece linealmente tras pasar las 200h

# Ganancias: A=$50, B=$40, C=$60
# Solo calculamos ganancia donde x3 >= 0 (puntos factibles)
ganancias = []
for h in horas_empaque:
    # Calculamos x1, x2, x3 para cada H usando Cramer
    d1 = (200*7 - 1*450 + 2*(150-3*200)) # Ejemplo de estructura
    # Simplificando la ganancia para la curva:
    if h < 200:
        # En la zona "ahogada", la ganancia es teórica (no real)
        g = 50*106 + 40*28 + 60*((h-200)/1) 
    else:
        # Zona factible
        val_x3 = (h - 200)
        val_x1 = 90 - (0.4 * val_x3) # A disminuye para dar espacio a C
        val_x2 = 20 + (0.2 * val_x3) # B aumenta ligeramente
        g = 50*val_x1 + 40*val_x2 + 60*val_x3
    ganancias.append(g)

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(horas_empaque, ganancias, color='green', linewidth=2, label='Ganancia Total ($)')
plt.axvline(x=200, color='red', linestyle='--', label='Punto de Equilibrio (C=0)')
plt.fill_between(horas_empaque, ganancias, where=(horas_empaque < 200), color='red', alpha=0.1)
plt.fill_between(horas_empaque, ganancias, where=(horas_empaque >= 200), color='green', alpha=0.1)

plt.title('Sensibilidad de la Ganancia vs Horas de Empaque')
plt.xlabel('Disponibilidad de Horas en Empaque')
plt.ylabel('Ganancia Total ($)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.annotate('Zona no factible (C < 0)', xy=(185, 4200), color='red')
plt.annotate('Zona de Optimización', xy=(240, 5500), color='darkgreen')
plt.show()
```
