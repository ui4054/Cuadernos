### Banco de Ejercicios: Sistemas de Ecuaciones 3x3 en Producción

---

**1. La Fábrica de Muebles "Roble Alto"**
Una carpintería produce sillas, mesas y armarios. Cada producto pasa por corte, ensamblaje y acabado. Una silla requiere 1h de corte, 2h de ensamblaje y 1h de acabado. Una mesa requiere 2h de corte, 1h de ensamblaje y 1h de acabado. Un armario requiere 3h de corte, 2h de ensamblaje y 2h de acabado. Si se dispone de 80h de corte, 70h de ensamblaje y 50h de acabado:
*   **Solución:** Sillas (x) = 10, Mesas (y) = 20, Armarios (z) = 10.

**2. Química "Solventes S.A."**
Se deben mezclar tres compuestos (A, B y C) para obtener tres tipos de fertilizantes. El fertilizante 1 usa 2 unidades de A, 1 de B y 1 de C. El fertilizante 2 usa 1 unidad de A, 2 de B y 1 de C. El fertilizante 3 usa 1 de A, 1 de B y 2 de C. Se tienen en inventario 100 unidades de A, 110 de B y 120 de C.
*   **Solución:** F1 = 22.5, F2 = 32.5, F3 = 42.5.

**3. "Tecnología Global" y los Microchips**
Se fabrican tres modelos de procesadores: Alpha, Beta y Gamma. El modelo Alpha requiere 10mg de silicio, 2mg de cobre y 1mg de oro. El Beta requiere 15mg de silicio, 3mg de cobre y 2mg de oro. El Gamma requiere 20mg de silicio, 5mg de cobre y 3mg de oro. La empresa dispone de 4500mg de silicio, 1050mg de cobre y 650mg de oro.
*   **Solución:** Alpha = 150, Beta = 100, Gamma = 75.

**4. Panificadora "El Trigo Dorado"**
Se producen tres tipos de panes especiales. El tipo 1 requiere 500g de harina, 50g de azúcar y 10g de levadura. El tipo 2 requiere 400g de harina, 60g de azúcar y 15g de levadura. El tipo 3 requiere 600g de harina, 40g de azúcar y 20g de levadura. Se consumieron 25kg de harina, 2.6kg de azúcar y 0.75kg de levadura.
*   **Solución:** Tipo 1 = 20, Tipo 2 = 15, Tipo 3 = 15.

**5. "Textiles Modernos"**
Una fábrica hace camisas, pantalones y chaquetas. Una camisa usa 1m de tela, 2 botones y 1 cremallera. Un pantalón usa 1.5m de tela, 1 botón y 1 cremallera. Una chaqueta usa 2m de tela, 5 botones y 2 cremalleras. El inventario gastado fue de 155m de tela, 210 botones y 110 cremalleras.
*   **Solución:** Camisas = 30, Pantalones = 30, Chaquetas = 40.

**6. Ensambladora de Bicicletas "Rueda Libre"**
Se ensamblan modelos de Montaña, Carreras y Urbana. La de Montaña requiere 3h de soldadura, 2h de pintura y 1h de ajuste. La de Carreras requiere 2h de soldadura, 2h de pintura y 2h de ajuste. La Urbana requiere 1h de soldadura, 2h de pintura y 3h de ajuste. Se dispone de 40h de soldadura, 44h de pintura y 48h de ajuste.
*   **Solución:** Montaña = 8, Carreras = 6, Urbana = 4.

**7. Concesionario "Autos Premium"**
Se preparan tres modelos para entrega: Sedán, SUV y Deportivo. El Sedán requiere 2h de lavado, 1h de pulido y 1h de revisión. La SUV requiere 3h de lavado, 2h de pulido y 1h de revisión. El Deportivo requiere 2h de lavado, 2h de pulido y 2h de revisión. El tiempo total fue de 64h de lavado, 44h de pulido y 36h de revisión.
*   **Solución:** Sedán = 10, SUV = 12, Deportivo = 4.

**8. Empresa de Catering "Sabores"**
Se preparan bandejas Ejecutiva, Premium y Deluxe. La Ejecutiva tiene 2 entradas, 3 platos fuertes y 1 postre. La Premium tiene 3 entradas, 4 platos fuertes y 2 postres. La Deluxe tiene 4 entradas, 5 platos fuertes y 3 postres. Se sirvieron en total 290 entradas, 400 platos fuertes y 190 postres.
*   **Solución:** Ejecutiva = 30, Premium = 50, Deluxe = 20.

**9. "Juguetes Creativos"**
Se fabrican robots, trenes y aviones de bloques. El robot usa 20 bloques rojos, 10 azules y 5 amarillos. El tren usa 15 rojos, 20 azules y 10 amarillos. El avión usa 10 rojos, 15 azules y 20 amarillos. Se usaron 1450 bloques rojos, 1450 azules y 1100 amarillos.
*   **Solución:** Robots = 40, Trenes = 30, Aviones = 20.

**10. "Logística Express"**
Se envían paquetes pequeños, medianos y grandes. El pequeño ocupa 1 pie cúbico y pesa 2kg. El mediano ocupa 3 pies cúbicos y pesa 5kg. El grande ocupa 5 pies cúbicos y pesa 10kg. Un contenedor de 100 pies cúbicos y 210kg de capacidad se llenó con 30 paquetes totales. ¿Cuántos de cada uno hay?
*   **Solución:** Pequeños = 10, Medianos = 10, Grandes = 10.
