### 1. La Fábrica de Muebles "Roble Alto"

**Planteamiento:**
- x: sillas, y: mesas, z: armarios
- Corte: `1x + 2y + 3z = 80`
- Ensamblaje: `2x + 1y + 2z = 70`
- Acabado: `1x + 1y + 2z = 50`

**Solución Correcta:**
- Resolviendo el sistema, se obtiene: x = 10, y = 20, z = 10.

**Revisión de la Solución del Cuaderno (10, 20, 10):**
- Ecuación de Corte: `1(10) + 2(20) + 3(10) = 10 + 40 + 30 = 80`. (Correcto)
- Ecuación de Ensamblaje: `2(10) + 1(20) + 2(10) = 20 + 20 + 20 = 60`. (INCORRECTO, debería ser 70)
- Ecuación de Acabado: `1(10) + 1(20) + 2(10) = 10 + 20 + 20 = 50`. (Correcto)
- **Conclusión:** El problema está mal formulado en el cuaderno. La solución solo es válida si las horas de ensamblaje fueran 60.
